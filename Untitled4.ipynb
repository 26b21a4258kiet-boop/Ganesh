```python

# ============================================================
# FACTORY PRODUCTION PLANNER (A7)
# B.Tech Mathematics Mini Project
# Concept: Consistency, Rank Analysis and Practical Interpretation
# ============================================================

"""
Factory Production Planner

This project models a factory that produces 3 products
using 3 machines.

Mathematical model:

        A x = b

where:
A = machine-hours required for each product
x = production quantities
b = available machine hours

The project checks:
1. Consistency of the system
2. Rank(A)
3. Rank([A | b])
4. Production quantities
5. Practical interpretation
6. Machine-hour utilization
7. Approximate solution when the system is inconsistent
"""
```

```python

# ============================================================
# IMPORTS
# ============================================================

import numpy as np
import sympy as sp
import matplotlib.pyplot as plt

# Optional: make NumPy output easier to read
np.set_printoptions(precision=3, suppress=True)
```

```python

# ============================================================
# USER CONFIGURATION
# ============================================================

# Names of products
products = ["Product A", "Product B", "Product C"]

# Names of machines
machines = ["Machine 1", "Machine 2", "Machine 3"]

# ------------------------------------------------------------
# Machine-hours required for ONE unit of each product
#
# Rows    -> Machines
# Columns -> Products
#
# Example:
# Machine 1 requires:
#   2 hours for Product A
#   1 hour  for Product B
#   1 hour  for Product C
# ------------------------------------------------------------

machine_hours = np.array([
    [2, 1, 1],   # Machine 1
    [1, 2, 1],   # Machine 2
    [1, 1, 2]    # Machine 3
], dtype=float)

# Available hours for each machine
machine_capacity = np.array([
    80,   # Machine 1
    90,   # Machine 2
    70    # Machine 3
], dtype=float)

# ------------------------------------------------------------
# Basic validation
# ------------------------------------------------------------

if machine_hours.shape != (3, 3):
    raise ValueError("machine_hours must be a 3 x 3 matrix.")

if len(machine_capacity) != 3:
    raise ValueError("machine_capacity must contain 3 values.")

if len(products) != 3 or len(machines) != 3:
    raise ValueError("This first version requires exactly 3 products and 3 machines.")

if np.any(machine_hours < 0):
    raise ValueError("Machine-hour requirements cannot be negative.")

if np.any(machine_capacity < 0):
    raise ValueError("Machine capacities cannot be negative.")
```

```python

# ============================================================
# INPUT DISPLAY
# ============================================================

print("=" * 65)
print("FACTORY PRODUCTION PLANNER")
print("=" * 65)

print("\nMachine-hour requirement per unit:")

header = f"{'Machine':<15}" + "".join(f"{p:<15}" for p in products)
print(header)
print("-" * len(header))

for i, machine in enumerate(machines):
    row = f"{machine:<15}"
    for j in range(3):
        row += f"{machine_hours[i, j]:<15.2f}"
    print(row)

print("\nAvailable machine hours:")

for machine, capacity in zip(machines, machine_capacity):
    print(f"{machine:<15}: {capacity:.2f} hours")
```

```python

# ============================================================
# MATHEMATICAL MODEL
# ============================================================

A = sp.Matrix(machine_hours.astype(int))
b = sp.Matrix(machine_capacity.astype(int))

print("\nMATHEMATICAL MODEL")
print("-" * 65)

print("\nCoefficient matrix A:")
sp.pprint(A)

print("\nProduction vector x:")
print("x = [x1, x2, x3]^T")

print("\nCapacity vector b:")
sp.pprint(b)

print("\nSystem of equations:")

for i in range(3):
    equation = " + ".join(
        f"{int(machine_hours[i, j])}x{j+1}"
        for j in range(3)
        if machine_hours[i, j] != 0
    )

    print(
        f"{machines[i]}: {equation} = {int(machine_capacity[i])}"
    )
```

```python

# ============================================================
# CONSISTENCY AND RANK ANALYSIS
# ============================================================

# Augmented matrix [A | b]
augmented_matrix = A.row_join(b)

rank_A = A.rank()
rank_augmented = augmented_matrix.rank()

print("\n" + "=" * 65)
print("CONSISTENCY AND RANK ANALYSIS")
print("=" * 65)

print("\nMatrix A:")
sp.pprint(A)

print("\nAugmented matrix [A | b]:")
sp.pprint(augmented_matrix)

print(f"\nRank(A)       = {rank_A}")
print(f"Rank([A | b]) = {rank_augmented}")

# ------------------------------------------------------------
# Consistency theorem
#
# A system Ax = b is consistent when:
#
# rank(A) = rank([A | b])
# ------------------------------------------------------------

if rank_A == rank_augmented:
    consistent = True
    print("\nRESULT: The system is CONSISTENT.")

    if rank_A == 3:
        print("The system has a UNIQUE solution.")
    else:
        print("The system has INFINITELY MANY solutions.")

else:
    consistent = False
    print("\nRESULT: The system is INCONSISTENT.")
    print("There is no exact production plan satisfying all equations.")
```

```python

# ============================================================
# SOLUTION ENGINE
# ============================================================

print("\n" + "=" * 65)
print("PRODUCTION QUANTITY CALCULATION")
print("=" * 65)

# ------------------------------------------------------------
# Case 1: Unique solution
# ------------------------------------------------------------

if consistent and rank_A == 3:

    # SymPy solves the exact system
    solution = sp.linsolve((A, b))

    solution_tuple = list(solution)[0]

    production = np.array(
        [float(value) for value in solution_tuple]
    )

    print("\nUnique production solution:")

    for product, quantity in zip(products, production):
        print(f"{product:<15}: {quantity:.2f} units")


# ------------------------------------------------------------
# Case 2: Consistent but infinitely many solutions
# ------------------------------------------------------------

elif consistent and rank_A < 3:

    solution = sp.linsolve((A, b))

    print("\nThe system is consistent but has infinitely many solutions.")

    print("\nGeneral solution:")
    sp.pprint(solution)

    print(
        "\nFor a unique production plan, additional business constraints "
        "would be required."
    )


# ------------------------------------------------------------
# Case 3: Inconsistent system
# ------------------------------------------------------------

else:

    print("\nNo exact production quantities exist.")

    print(
        "Calculating a least-squares approximate production plan "
        "for diagnostic purposes..."
    )

    # Numerical least-squares solution
    production, residuals, _, _ = np.linalg.lstsq(
        machine_hours,
        machine_capacity,
        rcond=None
    )

    print("\nApproximate production quantities:")

    for product, quantity in zip(products, production):
        print(f"{product:<15}: {quantity:.2f} units")
```

```python

# ============================================================
# SOLUTION VERIFICATION
# ============================================================

print("\n" + "=" * 65)
print("SOLUTION VERIFICATION")
print("=" * 65)

if "production" in globals():

    calculated_hours = machine_hours @ production

    print("\nMachine-hour calculation:")

    print(
        f"{'Machine':<15}"
        f"{'Available':<15}"
        f"{'Required':<15}"
        f"{'Difference':<15}"
    )

    print("-" * 60)

    for i, machine in enumerate(machines):

        available = machine_capacity[i]
        required = calculated_hours[i]
        difference = available - required

        print(
            f"{machine:<15}"
            f"{available:<15.2f}"
            f"{required:<15.2f}"
            f"{difference:<15.2f}"
        )

    if consistent:
        print("\nThe calculated production quantities satisfy the equations.")
    else:
        print(
            "\nBecause the original system is inconsistent, "
            "the approximate solution cannot satisfy every equation exactly."
        )
```

```python

# ============================================================
# MACHINE UTILIZATION
# ============================================================

print("\n" + "=" * 65)
print("MACHINE UTILIZATION ANALYSIS")
print("=" * 65)

if "production" in globals():

    used_hours = machine_hours @ production

    utilization = (
        used_hours / machine_capacity
    ) * 100

    print(
        f"{'Machine':<15}"
        f"{'Used Hours':<15}"
        f"{'Capacity':<15}"
        f"{'Utilization':<15}"
    )

    print("-" * 60)

    for i, machine in enumerate(machines):

        print(
            f"{machine:<15}"
            f"{used_hours[i]:<15.2f}"
            f"{machine_capacity[i]:<15.2f}"
            f"{utilization[i]:<15.2f}%"
        )

    # Identify the machine with the highest utilization.
    bottleneck_index = np.argmax(utilization)

    print(
        f"\nHighest-utilization machine: "
        f"{machines[bottleneck_index]}"
    )

    print(
        "This machine is the bottleneck candidate based on "
        "utilization in the calculated production plan."
    )
```

```python

# ============================================================
# INCONSISTENCY DIAGNOSTIC
# ============================================================

if not consistent:

    print("\n" + "=" * 65)
    print("INCONSISTENCY DIAGNOSTIC")
    print("=" * 65)

    calculated_hours = machine_hours @ production

    residuals = machine_capacity - calculated_hours

    print(
        "\nResidual = Available hours - Approximate required hours"
    )

    print(
        f"{'Machine':<15}"
        f"{'Residual':<15}"
        f"{'Absolute mismatch':<20}"
    )

    print("-" * 55)

    for i, machine in enumerate(machines):

        print(
            f"{machine:<15}"
            f"{residuals[i]:<15.3f}"
            f"{abs(residuals[i]):<20.3f}"
        )

    largest_mismatch = np.argmax(np.abs(residuals))

    print(
        f"\nLargest numerical mismatch: "
        f"{machines[largest_mismatch]}"
    )

    print(
        "\nInterpretation:"
    )

    print(
        "The machine with the largest residual is flagged as the "
        "largest mismatch in the approximate model."
    )

    print(
        "This should be treated as a diagnostic indicator rather "
        "than proof of a physical bottleneck."
    )
```

```python

# ============================================================
# FINAL RESULT TABLE
# ============================================================

print("\n" + "=" * 65)
print("FINAL PROJECT RESULT")
print("=" * 65)

if consistent and rank_A == 3:

    print("\nStatus       : CONSISTENT")
    print("Solution     : UNIQUE")

    print("\nProduction Plan:")
    print("-" * 40)

    for product, quantity in zip(products, production):
        print(f"{product:<20}: {quantity:>8.2f} units")

    print("\nMathematical conclusion:")
    print(
        f"rank(A) = rank([A|b]) = {rank_A}, "
        "so the system has a unique solution."
    )

elif consistent:

    print("\nStatus       : CONSISTENT")
    print("Solution     : INFINITELY MANY")

    print(
        "\nAdditional production constraints are needed "
        "to select one production plan."
    )

else:

    print("\nStatus       : INCONSISTENT")
    print("Solution     : NO EXACT SOLUTION")

    print(
        "\nAn approximate production plan is shown only "
        "for diagnostic purposes."
    )
```

```python

# ============================================================
# BENCHMARK TEST 1: CONSISTENT FACTORY
# ============================================================

products = ["Steel Part A", "Steel Part B", "Steel Part C"]

machines = [
    "Cutting Machine",
    "Drilling Machine",
    "Finishing Machine"
]

machine_hours = np.array([
    [2, 1, 1],
    [1, 2, 1],
    [1, 1, 2]
], dtype=float)

machine_capacity = np.array([
    80,
    90,
    70
], dtype=float)

print("Benchmark 1 loaded.")
print("Expected result: a unique production solution.")
```

```python

# ============================================================
# BENCHMARK TEST 2: INCONSISTENT SYSTEM
# ============================================================

products = [
    "Product A",
    "Product B",
    "Product C"
]

machines = [
    "Machine 1",
    "Machine 2",
    "Machine 3"
]

# Here Machine 3's row is mathematically dependent
# on Machine 1 and Machine 2:
#
# Row 3 = Row 1 + Row 2
#
# But the available hours do NOT satisfy:
#
# b3 = b1 + b2
#
# 149 != 70 + 80
#
# Therefore the system becomes inconsistent.

machine_hours = np.array([
    [2, 1, 1],
    [1, 2, 1],
    [3, 3, 2]
], dtype=float)

machine_capacity = np.array([
    70,
    80,
    149
], dtype=float)

print("Benchmark 2 loaded.")
print("Expected result: inconsistent system.")
```

```python

# ============================================================
# COMPLETE ANALYSIS FUNCTION
# ============================================================

def analyze_factory(products, machines, machine_hours, machine_capacity):
    """
    Complete mathematical analysis of the factory model.

    Model:
        A x = b

    Returns:
        A dictionary containing ranks, consistency,
        production solution and machine utilization.
    """

    # Convert inputs
    A_np = np.array(machine_hours, dtype=float)
    b_np = np.array(machine_capacity, dtype=float)

    # SymPy matrices for exact rank analysis
    A_sp = sp.Matrix(A_np.astype(int))
    b_sp = sp.Matrix(b_np.astype(int))

    augmented = A_sp.row_join(b_sp)

    # Rank calculation
    rank_A = A_sp.rank()
    rank_aug = augmented.rank()

    consistent = rank_A == rank_aug

    result = {
        "rank_A": rank_A,
        "rank_augmented": rank_aug,
        "consistent": consistent
    }

    # --------------------------------------------------------
    # Unique solution
    # --------------------------------------------------------

    if consistent and rank_A == A_sp.shape[1]:

        x = np.linalg.solve(A_np, b_np)

        result["solution_type"] = "Unique"
        result["production"] = x

    # --------------------------------------------------------
    # Infinite solutions
    # --------------------------------------------------------

    elif consistent:

        result["solution_type"] = "Infinite"
        result["production"] = None

    # --------------------------------------------------------
    # Inconsistent
    # --------------------------------------------------------

    else:

        x, _, _, _ = np.linalg.lstsq(
            A_np,
            b_np,
            rcond=None
        )

        result["solution_type"] = "Inconsistent"
        result["production"] = x

    # --------------------------------------------------------
    # Utilization if production is available
    # --------------------------------------------------------

    if result["production"] is not None:

        used = A_np @ result["production"]

        utilization = (
            used / b_np
        ) * 100

        result["used_hours"] = used
        result["utilization"] = utilization

    return result
```

```python

# ============================================================
# RUN COMPLETE ANALYSIS
# ============================================================

result = analyze_factory(
    products,
    machines,
    machine_hours,
    machine_capacity
)

print("=" * 65)
print("FACTORY PRODUCTION PLANNER — FINAL ANALYSIS")
print("=" * 65)

print(f"\nRank(A)       : {result['rank_A']}")
print(f"Rank([A | b]) : {result['rank_augmented']}")

if result["consistent"]:
    print("\nConsistency   : CONSISTENT")
else:
    print("\nConsistency   : INCONSISTENT")

print(f"Solution type : {result['solution_type']}")

if result["production"] is not None:

    print("\nProduction quantities:")

    for product, quantity in zip(
        products,
        result["production"]
    ):
        print(
            f"  {product:<20} {quantity:>8.2f} units"
        )

    print("\nMachine utilization:")

    for machine, utilization in zip(
        machines,
        result["utilization"]
    ):
        print(
            f"  {machine:<20} {utilization:>8.2f}%"
        )
```

```python

# ============================================================
# PRACTICAL INTERPRETATION
# ============================================================

print("\n" + "=" * 65)
print("PRACTICAL INTERPRETATION")
print("=" * 65)

if result["consistent"] and result["solution_type"] == "Unique":

    print("""
The factory planning equations are consistent.

The rank of the coefficient matrix equals the rank
of the augmented matrix.

Since the rank is equal to the number of unknowns,
there is one unique production plan.

Therefore, the calculated quantities tell the factory
how many units of each product can be produced while
exactly satisfying the three machine-hour equations.
""")

elif result["consistent"] and result["solution_type"] == "Infinite":

    print("""
The factory equations are consistent, but there are
infinitely many possible production plans.

The mathematical information is not sufficient to
select one unique plan.

Additional real-world constraints could be added,
such as:

- minimum production quantities
- maximum demand
- product priorities
- profit per product
- raw-material limits
""")

else:

    print("""
The factory equations are inconsistent.

Therefore, no production quantities can satisfy all
three machine-hour equations exactly at the same time.

This means the stated machine-hour requirements and
available hours cannot all be true simultaneously under
the full-utilization assumption.

The least-squares solution is only an approximate
diagnostic solution.

A real factory would normally reformulate this as a
capacity-constrained optimization problem using
inequalities such as:

    Ax <= b

rather than requiring every machine to be used exactly.
""")
```
